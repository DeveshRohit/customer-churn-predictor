#### Splitting data

```python
X = df.drop('Churn', axis=1)
y = df['Churn']

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y)
```

#### the stratify parameter ensures that the training and testing datasets maintain the same proportion of class labels as the original dataset

```python
median = X_train['TotalCharges'].median()

X_train['TotalCharges'] = X_train['TotalCharges'].fillna(median)
X_test['TotalCharges'] = X_test['TotalCharges'].fillna(median)
```

Only using the median from the train set, we don't use information from the test set

### SMOTE

The dataset is imbalanced, favouring the non-churn customers

```python
df['Churn'].value_counts()
```

As we can see,
- ~5000 non-churn customers
- ~2000 churning customers

But in reality we care for our churning customers. A model should predict which customers have a tendency to churn, so that loss can be prevented.<br>

Training on this dataset will give a model that will just classify maximum customers to be churning and give accuracy around 75%

### SMOTE: Synthetic Minority Over-sampling Technique
SMOTE creates new synthetic minority examples, and training the model on this would give a better performance<br>

Suppose there are two minority-class customers:<br>
A = [10, 50]
B = [20, 70]<br>

SMOTE can create a new customer somewhere between them.<br>
[10, 50] ─────────── [13, 56] ─────────── [20, 70] ([13, 56] created by some mathematical formula)

```python
numerical_cols = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]
```

```python
categorical_cols = [
    "gender",
    "SeniorCitizen",
    "Partner",
    "Dependents",
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaperlessBilling",
    "PaymentMethod"
]
```

The below cell is basically to tell the smotenc model that which cols are categorical and which are not

```python
categorical_indices = [
    X_train.columns.get_loc(col) for col in categorical_cols
]
```

Before applying smotenc, we must scale the continuous numerical columns

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaler.fit(X_train[numerical_cols])

X_train[numerical_cols] = scaler.transform(X_train[numerical_cols])

X_test[numerical_cols] = scaler.transform(X_test[numerical_cols])
```

```python
categorical_indices = [
    X_train.columns.get_loc(col)
    for col in categorical_cols
]

print(categorical_indices)
print(X_train.columns)
```

```python
# before smotenc

print(y_train.value_counts())
```

```python
from imblearn.over_sampling import SMOTENC

smotenc = SMOTENC(
    categorical_features=categorical_indices,
    random_state=42
)

X_train_resampled, y_train_resampled = smotenc.fit_resample(
    X_train,
    y_train
)
```

```python
# after smotenc

print(y_train_resampled.value_counts())
```

```python
X_train_resampled.head()
```

OHE on the resampled dataset

```python
X_train_resampled = pd.get_dummies(
    X_train_resampled,
    columns=categorical_cols,
    drop_first=True
)

X_test = pd.get_dummies(
    X_test,
    columns=categorical_cols,
    drop_first=True
)
```

However, the train and test sets might not have the same columns, so fixing that:

This fix is done so that both columns have the same columns, because by chance for e.g. the test set may not have a single credit card customer, that would simply omit the entire `PaymentMethod_Credit_card` column in the test set. That would be problematic.<br>
`fill_value=0` because if the column does not exist, that means there are no credit card customers (0 in OHE terms).

```python
X_train_resampled, X_test = X_train_resampled.align(
    X_test,
    join="left",    # X_train_resampled is the superset
    axis=1,
    fill_value=0
)
```

Checking the final data

```python
print(X_train_resampled.shape)
print(X_test.shape)
```

```python
print(
    set(X_train_resampled.columns) - set(X_test.columns)
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

### LOGISTIC REGRESSION

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(random_state=42)

lr.fit(X_train_resampled, y_train_resampled)
```

```python
y_pred = lr.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1:", f1_score(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

### DECISION TREE

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(random_state=42)

dt.fit(X_train_resampled, y_train_resampled)

y_pred_dt = dt.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_dt))
print("Precision:", precision_score(y_test, y_pred_dt))
print("Recall:", recall_score(y_test, y_pred_dt))
print("F1:", f1_score(y_test, y_pred_dt))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_dt))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_dt))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    random_state=42,
    n_estimators=200,
    n_jobs=-1
)

rf.fit(X_train_resampled, y_train_resampled)

y_pred_rf = rf.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_rf))
print("Precision:", precision_score(y_test, y_pred_rf))
print("Recall:", recall_score(y_test, y_pred_rf))
print("F1:", f1_score(y_test, y_pred_rf))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python
from sklearn.metrics import roc_auc_score

y_prob_rf = rf.predict_proba(X_test)[:, 1]

print("ROC-AUC:", roc_auc_score(y_test, y_prob_rf))
```

**What 82.25% ROC-AUC means**<br>
If you randomly pick one customer who churned and one who didn't, there's about an 82.25% chance that the model assigns the churner a higher churn probability

### **Observations**

It may look like RF > LR in terms of accuracy<br>

However, one key point that we must not overlook is the *recall* of Churners (1) of both the models, which is significantly higher in LR (~78%) compared to RF (~61%)

### Keeping in mind that ACCURACY is still not being considered a reliable metric 

```python
y_prob_lr = lr.predict_proba(X_test)[:, 1]

print("ROC-AUC for LR:", roc_auc_score(y_test, y_prob_lr))
```

```python
X = df.drop("Churn", axis=1)
y = df["Churn"]

X_train_no_smote, X_test_no_smote, y_train_no_smote, y_test_no_smote = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)
```

```python
median = X_train_no_smote["TotalCharges"].median()

X_train_no_smote["TotalCharges"] = X_train_no_smote["TotalCharges"].fillna(median)
X_test_no_smote["TotalCharges"] = X_test_no_smote["TotalCharges"].fillna(median)
```

```python
scaler_no_smote = StandardScaler()

scaler_no_smote.fit(X_train_no_smote[numerical_cols])

X_train_no_smote[numerical_cols] = scaler_no_smote.transform(
    X_train_no_smote[numerical_cols]
)

X_test_no_smote[numerical_cols] = scaler_no_smote.transform(
    X_test_no_smote[numerical_cols]
)
```

```python
X_train_no_smote = pd.get_dummies(
    X_train_no_smote,
    columns=categorical_cols,
    drop_first=True
)

X_test_no_smote = pd.get_dummies(
    X_test_no_smote,
    columns=categorical_cols,
    drop_first=True
)
```

I am also doing experiments with thresholds, changing the threshold might have an impact on the model's performance.

```python
thresholds = np.arange(0.2, 0.61, 0.05)

for threshold in thresholds:
    y_pred_threshold = (y_prob_lr_no_smote >= threshold).astype(int)

    print(
        f"Threshold: {threshold:.2f} | "
        f"Precision: {precision_score(y_test, y_pred_threshold):.3f} | "
        f"Recall: {recall_score(y_test, y_pred_threshold):.3f} | "
        f"F1: {f1_score(y_test, y_pred_threshold):.3f}"
    )
```

Just focusing on recall doesn't mean we can ignore the other metrics.<br>
So, I am keeping the cutoff of recall to atleast 75%, and then experimenting with different thresholds to see which threshold gives the best precision for >75% recall.

```python
thresholds = np.arange(0.01, 0.51, 0.001)

results = []

for threshold in thresholds:
    y_pred = (y_prob_lr_no_smote >= threshold).astype(int)

    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)

    results.append({
        'threshold': threshold,
        "precision": precision,
        "recall": recall,
        "f1": f1
    })

threshold_df = pd.DataFrame(results)

best = (
    threshold_df[threshold_df['recall'] >= 0.75].sort_values('precision', ascending=False).iloc[0]
)

print(best)
```
